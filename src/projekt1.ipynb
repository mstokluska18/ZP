    przechowywać aktualną wartość licznika,
    pozwalać zwiększyć wartość o 1,
    pozwalać zmniejszyć wartość o 1,

    pozwalać odczytać aktualną wartość.

Na końcu napisz krótki program testujący, który tworzy licznik i wykonuje na nim kilka operacji.

```python
class Counter:
    def __init__(self):
        self.value = 0
        print('Creating a Counter')
    
    def increase(self):
        self.value +=1

    def decrease(self):
        self.value -=1

    def reset(self):
        self.value = 0
    
    def get_value(self):
        return self.value

#do testowania licznika
licznik1 = Counter()
```

```python
licznik1.increase()
print(f'Wartość licznika po:{licznik1.get_value()}')
licznik1.increase()
print(f'Wartość licznika po:{licznik1.get_value()}')
```

```python
licznik1.decrease()
print(f'Wartość licznika po:{licznik1.get_value()}')
```

```python
licznik1.reset()
print(f'Wartość licznika po:{licznik1.get_value()}')
```

Napisz klasę Vector2D, która przechowuje współrzędne x i y.

Klasa powinna umożliwiać:

    utworzenie wektora o zadanych współrzędnych,
    wypisanie go w czytelnej postaci,
    dodawanie dwóch wektorów operatorem +,
    obliczenie długości wektora,
    mnożenie wektora przez liczbę.



```python
import math
class Vector2D:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __str__(self):
        return (f'Wspołrzędne wektora: x={self.x}, y={self.y}')

    def __add__(self, other):
        nowy_x=self.x + other.x
        nowy_y=self.y + other.y
        return Vector2D(nowy_x, nowy_y)

    def __mul__(self, number):
        nowy_x = self.x * number
        nowy_y = self.y * number
        return Vector2D(nowy_x, nowy_y)

    __rmul__ = __mul__ #5*wektor = wektor*5

    def length(self):
        return math.sqrt(self.x**2 + self.y**2)

v1=Vector2D(30,20)
v2=Vector2D(10,5)

print('Wektor v1:',v1,'jego długość:',v1.length())
print('Wektor v2:',v2,'jego długość:',v2.length())

v3=v1+v2
print('Wektor v3 (v1+v2):',v3,'jego długość:',v3.length())

v4=v1*(0.5)
print('Wektor v4 (v1*0.5):',v4,'jego długość:',v4.length())

        
```

    obiekt funkcyjny

Rozszerzenie

Dodać metodę zwracającą wartość pochodnej wielomianu w punkcie.

```python
class Polynomial:
    def __init__(self, wspolczynniki):
        self.wspolczynniki = wspolczynniki #lista wspołczynników

    def __str__(self):
        elementy=[]
        for potega, wspolczynnik in enumerate(self.wspolczynniki):
           elementy.append(f'{wspolczynnik}*x^{potega}')
        return '+'.join(elementy)#join skleja stringi

    #__call__ dla obiektu wywołuje instrukcje zdefiniowane w call

    def __call__(self, x):
        wynik =0
        for potega, wspolczynnik in enumerate(self.wspolczynniki):
            wynik += wspolczynnik * x**potega
        return wynik

    def get_degree(self):
        if len(self.wspolczynniki)== 0:
            return 0
        return len(self.wspolczynniki) -1

    def pochodna_w_punkcie(self, x):
        wynik =0
        for potega, wspolczynnik in enumerate(self.wspolczynniki):
            if potega > 0:
                wynik += potega * wspolczynnik * x**(potega -1)
        return wynik




wielomian = Polynomial([1, 0, 3]) # wielomian: 1 + 0x + 3x^2
print(f'Dla x=2 wynik to: {wielomian(2)}')

wynik_pochodnej = wielomian.pochodna_w_punkcie(3)
print(f"Pochodna w punkcie x=3 wynosi: {wynik_pochodnej}") 
```

Następnie napisz klasę PremiumAccount, dziedziczącą po Account, która dodatkowo:

    przechowuje współczynnik premii,
    przy dodawaniu punktów dolicza bonus,
    korzysta z super() tam, gdzie ma to sens.

Można się nauczyć

    dziedziczenie
    nadpisywanie metod
    super()

Rozszerzenie

Dodać trzeci typ konta, np. PenaltyAccount, które przy każdej operacji pobiera niewielką opłatę w punktach.

```python
class Account:
    def __init__(self,nazwa): #gdyby sienie zerowaly punkty to w nawiasie punkty i na dole self.puntky-punkty
        self.nazwa = nazwa
        self.punkty = 0    
    
    def add_points(self, points):
        self.punkty += points
    
    def remove_points(self, points):
        self.punkty -= points

    def info(self):
        print(f'Konto: {self.nazwa}, Punkty: {self.punkty}')

class PremiumAccount(Account):
    def __init__(self,nazwa,premia):
        super().__init__(nazwa)
        self.premia=premia
    
    def add_points(self, points):
        suma = points + self.premia
        super().add_points(suma)

class PenaltyAccount(Account):
    def __init__(self, name, fee):
        super().__init__(name)
        self.fee = fee

    def add_points(self,points):
        ostateczne_punkty = points -self.fee
        super().add_points(ostateczne_punkty)
    
    def remove_points(self,points):
        ostateczne_punkty = points +self.fee
        super().remove_points(ostateczne_punkty)
    

#zwykle konto
zwykle = Account('Adam')
zwykle.add_points(100)
zwykle.remove_points(20)
zwykle.info()

#premium
premium = PremiumAccount('Anna',premia=100)
premium.add_points(10)
premium.info()
premium.remove_points(50)
premium.info()

#penalty
penalty = PenaltyAccount('Elsa', fee=5)
penalty.add_points(50)
penalty.info()
penalty.remove_points(20)
penalty.info()  
```

Rozszerzenie

Niech klasa Animal będzie zaimplementowana jako abstrakcyjna.

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    def __init__(self, imie, x, y):  
        self.imie = imie
        self.x = x
        self.y = y
        pass

    def info(self):
        print(f'Imię: {self.imie}, Pozycja: ({self.x}, {self.y})')
        pass

    @abstractmethod
    def sound(self):
        pass

    @abstractmethod
    def move(self):
        pass

class Dog(Animal):
    def __init__(self, imie, x, y):
        super().__init__(imie, x, y)

    def sound(self):
        print('Hau hau!')
        pass

    def move(self):
        self.x += 2
        print('Pies się porusza.')
        pass

class Cat(Animal):
    def __init__(self, imie, x, y):
        super().__init__(imie, x, y)

    def sound(self):
        print('Miau miau!')
        pass

    def move(self):
        self.y += 1
        print('Kot skacze.')
        pass

pies = Dog('Reksio', 0, 0)
kot = Cat('Mruczek', 0, 0)

pies.info()
pies.sound()    
pies.move() 
pies.info()     

kot.info()
kot.sound()
kot.move()
kot.info()  
```
