```python
import random

# dodajmy metodę
class SmartPhone:
    def __init__(self, name = ''):
        print('Creating a SmartPhone')
        self.name = name
        self.address_book = {}
        self._secret_code = random.randint(1000, 9999) # losowy kod, który będzie tajny dla każdego telefonu
        # jezeli atrybut jest z podloga, to nie ruszac tego atrybutu, bo jest on prywatny i nie powinien być modyfikowany z zewnątrz klasy
        
    def restart(self): # musi przyjac argument self jako pierwszy argument
        print(f'Restarting {self.name}')
        
    def send_data_to_china(self):
        print(f'Sending data from {self._secret_code} to China')
        
    def __str__(self):# metoda, która zwraca ładną reprezentację obiektu, gdy wywołamy print na obiekcie klasy SmartPhone
        return f'SmartPhone(name={self.name}, address_book={self.address_book})' # zwraca nazwe telefonu i jego książkę adresową, gdy wywołamy print na obiekcie klasy SmartPhone
    
    def __getitem__(self, key): # wywoluje sie, gdy próbujemy uzyskać dostęp do elementu za pomocą indeksu, np. spA['Bob']
        return self.address_book[key]
    
    def __setitem__(self, key, value): # wywołuje się, gdy próbujemy ustawić wartość elementu za pomocą indeksu, np. spA['Bob'] = '555-555-5555'
        self.address_book[key] = value
        
    def __len__(self): # zwraca liczbe wpisów w książce adresowej, gdy wywołamy len(spA)
        return len(self.address_book)
```

#DZIEDZICZENIE

```python
class ElectronicDevice:
    def __init__(self,name=''): #konstruktor, uruchamai się automatycznie gdy tworzymy nowy telefon, domyślny pusta nazwa teelfonu ''
        print('creating device')
        self.name=name
    def restart(self):
        print(f'restarting{self.name}')

class Computer(ElectronicDevice): #dziedziczenie, czyli klasa Computer dziedziczy po klasie ElectronicDevice
    pass
```

class Computer(ElectronicDevice): – Mówisz Pythonowi: „Komputer to rodzaj urządzenia elektronicznego, ma mieć wszystko to samo, co ono”.

    pass – Oznacza: „nie dodaję nic nowego do komputera, użyj dokładnie tego, co jest w przepisie na urządzenie”.

    Dziedziczenie konstruktora: Ponieważ Computer nie ma własnego __init__, Python idzie „piętro wyżej” do ElectronicDevice i tam szuka instrukcji. Jeśli tam znajdzie poprawnie zapisane __init__, to go użyje.

```python
mypc = Computer('My PC') # wywołuje się konstruktor klasy bazowej
```

```python
mypc.restart()# wywołuje metodę restart, która jest dziedziczona z klasy ElectronicDevice, więc mypc może korzystać z tej metody, mimo że jest obiektem klasy Computer
```

uruchamia funkcję (metodę), która została zdefiniowana w klasie nadrzędne
Szukanie metody: Python zagląda do obiektu mypc i pyta: „Czy masz funkcję restart?”.

Mechanizm dziedziczenia: Ponieważ Computer dziedziczy po ElectronicDevice, Python sprawdza przepis klasy bazowej i tam znajduje definicję tej metody.

Wykonanie kodu: Odpala się linijka print(f'restarting {self.name}').

Użycie danych obiektu: self.name zostanie zastąpione nazwą, którą podałeś przy tworzeniu komputera (czyli 'My PC').

### MRO - Method Resolution Order - Kolejność Rozwiązywania Metod

Kiedy wywołujesz mypc.restart(), Python nie szuka na oślep. Idzie ścieżką MRO:

    Klasa obiektu: Najpierw sprawdza, czy restart jest w klasie Computer.

    Klasa nadrzędna: Jeśli nie ma go w Computer, idzie do ElectronicDevice.

    Klasa object: Na samym końcu zawsze sprawdza wbudowaną klasę object (to „pradziadek” wszystkich klas w Pythonie).

```python
#MRO metod resolution order
Computer.mro()
```

# 1. KLASA I INSTANCJA:
# Klasa (np. SmartPhone) to "przepis". 
# Instancja/Obiekt (np. spB) to konkretny telefon stworzony z tego przepisu.

# 2. KONSTRUKTOR __init__:
# Magiczna metoda, która odpala się SAMA przy tworzeniu obiektu (np. spB = SmartPhone()).
# Służy do ustawienia startowych danych (atrybutów) obiektu.
# WAŻNE: Muszą być DWIE podłogi z każdej strony (__), inaczej nie zadziała.

# 3. SŁOWO KLUCZOWE "self":
# Reprezentuje "ten konkretny" obiekt, na którym właśnie pracujemy.
# Dzięki self.name = name, Python wie, czyją nazwę ustawia (Boba czy Alice).

# 4. DZIEDZICZENIE:
# Klasa "dziecko" (np. Computer) przejmuje wszystkie cechy "rodzica" (np. ElectronicDevice).
# Pozwala to nie pisać tego samego kodu dwa razy.
Dziedziczenie: Pozwala klasie potomnej (np. Computer) przejąć metody i atrybuty klasy nadrzędnej (np. ElectronicDevice). Zapisuje się to tak: class Computer(ElectronicDevice):

# 5. FUNKCJA super():
# Używamy jej w "dziecku", żeby wywołać kod z "rodzica" (najczęściej w konstruktorze).
# Bez super().__init__(name), dziecko "zapomni" ustawić atrybuty od rodzica, co wywali błąd.

# 6. MRO (Method Resolution Order):
# Kolejka, w której Python szuka metod (najpierw w dziecku, potem w rodzicach, na końcu w 'object').
NazwaKlasy.__mro__. Python najpierw szuka metody w klasie Computer, a jeśli jej nie znajdzie – idzie wyżej do ElectronicDevice

# 7. METODY SPECJALNE (DUNDER):
# __str__ -> co wypisać, gdy zrobimy print(obiekt).
# __len__ -> co zwrócić, gdy zapytamy o len(obiekt).
# __getitem__ / __setitem__ -> obsługa nawiasów kwadratowych [] jak w słowniku.

```python
class ElectronicDevice:
    def __init__(self, name = ''):
        print('Creating an Electronic Device')
        self.name = name
        
    def restart(self):
        print(f'Restarting {self.name}')
        
class Computer(ElectronicDevice): # dziedziczenie 
    def __init__(self, name, ram):
        print('Creating a Computer')
        super().__init__(name) # sprawdza jaka klasa jest kolejna w MRO, 
        # czyli pozwala wywoływać konstruktor klasy nadrzędnej
        self.ram = ram
```
